# Coroutines and Tasks
This section outlines high-level asyncio APIs to work with coroutines and Tasks.

```python
import numpy as np
```

```python
a = np.array([[0.51334993, 0.27276812],
                   [1.19196704, 1.45208249]])
```

```python
np.block(a)
```

```python

bloque = np.array([[0.51334993, 0.27276812],
                   [1.19196704, 1.45208249]])
bloque2 = np.array([[0.51334993, 0.27276812],
                    [1.19196704, 1.45208249]])
bloque3 = np.array([[0.51334993, 0.27276812],
                    [1.19196704, 1.45208249]])
bloque4 = np.array([[0.51334993, 0.27276812],
                    [1.19196704, 1.45208249]])

# Combinar los bloques en la forma correcta para np.block
C = np.block([[bloque, bloque2],
              [bloque3, bloque4]])
```
